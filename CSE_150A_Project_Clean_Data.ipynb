```python
import numpy as np
```

Separate each decision made by a player into data points.

```python
def read_row(row: str):
  sep = row.split(",")

  dealer_seen = int(sep[8])
  if (sep[8] == "1"):
    dealer_seen = 11

  val = 0
  num_aces = 0
  if sep[2] == "1" or sep[2] == "11":
    num_aces += 1
    val += 11
  else:
    val += int(sep[2])
  if sep[3] == "1" or sep[3] == "11":
    num_aces += 1
    val += 11
  else:
    val += int(sep[3])
  aces_used = 0

  ret = []
  if val >= 21:
    if val == 21:
      ret.append((dealer_seen, val, min(num_aces, 2), "STAND"))
      return ret
    else:
      val -= 10
      aces_used += 1

  for i in [4, 5, 6]:
    if sep[i] != "0":
      ret.append((dealer_seen, val, min(num_aces, 2), "HIT"))
      if sep[i] == "1" or sep[i] == "11":
        num_aces += 1
        val += 11
      else:
        val += int(sep[i])

      while val > 21 and aces_used < num_aces:
        val -= 10
        aces_used += 1

      if val == 21:
        ret.append((dealer_seen, val, min(num_aces, 2), "STAND"))
        return ret
      elif val > 21:
        return ret

  ret.append((dealer_seen, val, min(num_aces, 2), "STAND"))
  return ret
```

Open the file, then read each row of data into the function above.

```python
datum = np.ndarray([13, 23, 7, 2])
for dealer in range(2, 12):
  for player in range(1, 22):
    for num_aces in range(0, 6):
      datum[dealer][player][num_aces][0] = 0
      datum[dealer][player][num_aces][1] = 0

with open("blkjckhands.csv", 'r') as f:
    i = 0
    for row in f:
      i += 1
      if i == 1:
        continue

      # if i <= 20:
      for eenie, meenie, miney, moe in read_row(row):
        if moe == "HIT":
          datum[eenie][meenie][miney][0] += 1
        else:
          datum[eenie][meenie][miney][1] += 1

filename = 'blkjck_clean.csv'
with open(filename, 'w') as f:
    print('dealer_seen,player_value,num_aces,num_hits,num_stands', file=f)
    for dealer in range(2, 12):
      for player in range(4, 22):
        for num_aces in range(0, 3):
          print(f"{dealer},{player},{num_aces},{round(datum[dealer][player][num_aces][0])},{round(datum[dealer][player][num_aces][1])}", file=f)
```
